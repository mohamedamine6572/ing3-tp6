```python

# Import pandas library to work with data
import pandas as pd

# Create a list containing the 10 used cars
data = [
    ("Clio", 2017, 95000, "diesel", 1650000), # First Clio car
    ("Clio", 2019, 60000, "petrol", 1950000), # Second Clio car
    ("Clio", 2021, 25000, "petrol", 2350000), # Third Clio car
    ("Picanto", 2018, 70000, "petrol", 1550000), # First Picanto car
    ("Picanto", 2021, 30000, "petrol", 1900000), # Second Picanto car
    ("Symbol", 2016, 120000, "diesel", 1300000), # First Symbol car
    ("Symbol", 2020, 50000, "petrol", 1700000), # Second Symbol car
    ("Symbol", 2022, 20000, "diesel", 2100000), # Third Symbol car
    ("Tucson", 2015, 150000, "diesel", 2900000), # First Tucson car
    ("Tucson", 2019, 80000, "diesel", 3800000) # Second Tucson car
]

# Create a DataFrame with the five required columns
df = pd.DataFrame(data, columns=["model", "year", "km", "fuel", "price"])

# Q5: Select cars that have less than 70000 kilometers
cars_less_70000 = df[df["km"] < 70000]

# Count the number of selected cars
q5 = len(cars_less_70000)

# Print the number of cars
print("Q5:", q5)

# Q6: Select diesel cars and calculate their average price
diesel_avg = df[df["fuel"] == "diesel"]["price"].mean()

# Select petrol cars and calculate their average price
petrol_avg = df[df["fuel"] == "petrol"]["price"].mean()

# Print both average prices on one line
print(f"Q6: diesel {diesel_avg:.0f}, petrol {petrol_avg:.0f}")

# Q7: Calculate the difference between the highest and lowest price for each model
price_gap = df.groupby("model")["price"].max() - df.groupby("model")["price"].min()

# Find the model with the biggest price difference
biggest_gap_model = price_gap.idxmax()

# Print the model and its price difference
print(f"Q7: {biggest_gap_model} {price_gap[biggest_gap_model]}")

# Q8: Add a new column containing the price per kilometer
df["price_per_km"] = df["price"] / df["km"]

# Find the car with the lowest price per kilometer
best_value = df.loc[df["price_per_km"].idxmin()]

# Print the model and year of that car
print(f"Q8: {best_value['model']} {best_value['year']}")

```
