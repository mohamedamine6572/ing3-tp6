```python

#------------------------------Q1-4------------------------------

# Import pandas library to work with data
import pandas as pd

# Create a list containing the 10 used cars
data = [
    ("Clio", 2017, 95000, "diesel", 1650000), # First Clio car
    ("Clio", 2019, 60000, "petrol", 1950000), # Second Clio car
    ("Clio", 2021, 25000, "petrol", 2350000), # Third Clio car
    ("Picanto", 2018, 70000, "petrol", 1550000), # First Picanto car
    ("Picanto", 2021, 30000, "petrol", 1900000), # Second Picanto car
    ("Symbol", 2016, 120000, "diesel", 1300000), # First Symbol car
    ("Symbol", 2020, 50000, "petrol", 1700000), # Second Symbol car
    ("Symbol", 2022, 20000, "diesel", 2100000), # Third Symbol car
    ("Tucson", 2015, 150000, "diesel", 2900000), # First Tucson car
    ("Tucson", 2019, 80000, "diesel", 3800000) # Second Tucson car
]

# Create a DataFrame with the five required columns
df = pd.DataFrame(data, columns=["model", "year", "km", "fuel", "price"])

# Q1: Calculate the average price for each model
# Sort the average prices from cheapest to most expensive
q1 = df.groupby("model")["price"].mean().round().astype(int).sort_values()

# Print the model names and their average prices on one line
print("Q1:", ", ".join(f"{model} {price}" for model, price in q1.items()))

# Q2: Calculate the average kilometers for each model
q2 = df.groupby("model")["km"].mean()

# Find the model with the highest average kilometers
highest_km_model = q2.idxmax()

# Print the model with the highest average kilometers
print("Q2:", highest_km_model)

# Q3: Add a new column containing the age of each car
df["age"] = 2026 - df["year"]

# Select only diesel cars and calculate their average age
diesel_age = df[df["fuel"] == "diesel"]["age"].mean()

# Print the average age of diesel cars with one decimal
print(f"Q3: {diesel_age:.1f}")

# Q4: Select cars manufactured in 2020 or later
cars_2020 = df[df["year"] >= 2020]

# Find the cheapest car among the selected cars
cheapest = cars_2020.loc[cars_2020["price"].idxmin()]

# Print the model, year and price of the cheapest car
print(f"Q4: {cheapest['model']} {cheapest['year']} {cheapest['price']}")
```
